```python
from sqlmodel import create_engine, Session
from sqlmodel import Field, SQLModel, Session, create_engine

class Song(SQLModel, table=True):
    song_id:int = Field(default=None, primary_key=True)
    song_title: str
    artist: str
    year: int | None = None

engine = create_engine(
    "postgresql://username:password@localhost/music_database",
    echo=True
)

SQLModel.metadata.create_all(engine)
with Session(engine) as session:
    song1 = Song(song_title = "無眠",artist="蘇打綠",year=2009)
    song2 = Song(song_title = "魚仔",artist="盧廣仲",year=2017)
    song3 = Song(song_title = "告白氣球",artist="周杰倫")
session.add(song1)
session.add(song2)
session.add(song3)

session.commit()
```
